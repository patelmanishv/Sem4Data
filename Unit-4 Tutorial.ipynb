## python based basic linear regression

### 1. IMPORT LIBRARIES

```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

### 2. IMPORT DATASET

```python
X  = np.array([60,62,67,70,71,72,75,78]).reshape(-1,1)
Y = np.array([140,155,159,179,192,200,212,215])
print(X)
print(Y)
print(X.shape)
print(Y.shape)
```

### 3. CREATE LINEAR REGRESSION MODEL

```python
model = LinearRegression()
```

### 4. FIT THE MODEL

```python
model.fit(X,Y)
```

### 5. GET SLOPE AND INTERCEPT

```python
slope = model.coef_[0]
intercept = model.intercept_
print(f"The slope is {slope} and intercept is {intercept}")
```

### 6. REGRESSION LINE EQUATION

```python
print(f"The linear regression equation is y = {intercept:.2f} + {slope:.2f}*x")
```

```python
# FOR X = 90 THE VALUE OF Y IS 272.7
model.predict([[90]])
```

## A TEMPLATE PROGRAM FOR SOLVING LINEAR REGRESSION (FOR EXAM)

## 1. Import libraries

```python
import numpy as np
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
import matplotlib.pyplot as plt
```

## 2. Load the dataset

```python
iris = datasets.load_iris()
x = iris.data[:,2].reshape(-1,1) # Petal Length
y = iris.data[:,3] # Petal width
print(x.shape)
print(y.shape)
```

## 3. SPLIT THE DATASET

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,random_state=42
)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

## 4. Create model

```python
model = LinearRegression()
```

## 5. Fit the model

```python
model.fit(x_train,y_train)
```

## PRINT SLOPE AND INTERCEPT 

```python
slope = model.coef_[0]
intercept = model.intercept_
print(f"Slope is {slope:.2f} and intercept is {intercept:.2f}")
```

## PREDICTION

```python
y_pred = model.predict(x_test)
print(y_pred)
```

## EVALUATION

```python
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)
print(f"Mean Squared Error is {mse:.2f} and R2 score is {r2:.2f}")
```

## PLOT THE REGRESSION LINE (OPTIONAL)

```python
plt.figure(figsize=(7,5))
plt.scatter(x_test,y_test,color = 'black',label="Actual Data")
plt.plot(x_test,y_pred,color='blue',label="Regressionline",linewidth=2)
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Linear Regression: PetalLength vs PetalWidth")
plt.legend()
x_min,x_max = plt.xlim()
y_min,y_max = plt.ylim()
equation = f"y = {intercept:.2f} + {slope:.2f}*x"
plt.text(
    x_max - 0.05 * (x_max - x_min),
    y_max + 0.05 * (y_max - y_min),
    equation,fontsize = 10,color='darkgreen',ha='right',
    bbox=dict(facecolor='white',alpha=0.8)  
)
plt.show()


```

## MULTIPLE LINEAR REGRESSION
- In multiple linear regression, there are multiple features or independent variables and one dependent variable (target)
  

```python
import pandas as pd
data = pd.DataFrame({
    'size': [2104,1600,2400,1416,3000],
    'year': [2003,1977,2001,1995,2000],
    'price':[399900,329900,369000,232000,539900]
})
```

```python
# Features and target
x = data[['size','year']]
y = data['price']
print(x.shape)
print(y.shape)
```

```python
from sklearn.linear_model import LinearRegression
# Create model
model = LinearRegression()
# Fit the model
model.fit(x,y)
# SLR Slope and intercept
print(model.coef_)
print(model.intercept_)
# R2score
print(model.score(x,y))
```

### NEW INPUT

```python
new_input = pd.DataFrame({
    'size':[750],
    'year':[2009]
})
prediction = model.predict(new_input)
print("Prediction for 750 Sq Ft area, 2009 is ",prediction[0])
```

## POLYNOMIAL REGRESSION

```python
# IMPORT LIBRARIES
import numpy as np
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt # Optional
import seaborn as sns
# Load the dataset 
iris = sns.load_dataset('iris')
iris.head()
iris.shape
# Define features and target
x = iris[['sepal_length']]
y = iris['petal_length']
print(x.shape)
print(y.shape)
# CREATE POLYNOMIAL FEATURES
poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x)
model = LinearRegression()
model.fit(x_poly,y)
print(model.coef_)
print(model.intercept_)
```

```python
x_test_values = np.linspace(4,8,100).reshape(-1,1)
x_test = pd.DataFrame(x_test_values,columns=['sepal_length'])
x_test_poly = poly.transform(x_test)
y_pred = model.predict(x_test_poly)
print(y_pred)
```

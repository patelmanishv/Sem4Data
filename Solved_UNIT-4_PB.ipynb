### 226

For the given RealEstate csv, write a python program satisfying following tasks to demonstrate application of machine learning through multiple linear regression as follows –

Dataset: RealEstate.csv
ML Library to be used: scikit-learn
Dependent variable: 'Y house price of unit area'
Independent variables: 'X1 transaction date', 'X2 house age', 'X3 distance to the nearest MRT station', 'X4 number of convenience stores', 'X5 latitude' and 'X6 longitude'

1. Import required libraries.
2. Load RealEstate dataset, create a dataframe and check datatypes of its attributes using appropriate method.
3. Remove ‘No’ column from the dataframe.
4. Check for any null values in features using appropriate method.
5. Create feature variables x and y as given above.
6. Create training and testing sets of feature variables with 70% of data for training and with random state of 110.
7. Create and fit regression model using appropriate method.
8. Use testing set created in step 6 to find and print the prediction of the outcome.
9. Find and print coefficient and mean squared error of the regression model.

---

```python
# 1. Import libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score

# 2. Load RealEstate dataset, create a dataframe and 
# check datatypes of its attributes using appropriate method.
df = pd.read_csv("RealEstate.csv")
df.info()

# Remove ‘No’ column from the dataframe.
df.head(2)
df.drop(columns=["No"],inplace=True)

df.head(2)

# Check for any null values in features using appropriate method.
df.isna().sum()

# Dependent variable: 'Y house price of unit area' 
# Independent variables: 'X1 transaction date', 'X2 house age', 'X3 distance to the nearest MRT station', 'X4 number of convenience stores', 'X5 latitude' and 'X6 longitude'
x = df.drop(columns=['Y house price of unit area'])
y = df['Y house price of unit area']
print(x.shape)
print(y.shape)

# Create training and testing sets of feature variables with 70% of data for training 
# and with random state of 110.
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.3,random_state=110
)

print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)

## Create and fit regression model using appropriate method.
model = LinearRegression()
model.fit(x_train,y_train)

# Use testing set created in step 6 to find 
# and print the prediction of the outcome.
prediction = model.predict(x_test)
print(prediction)

# Find and print coefficient and mean squared error of the regression model.
print(model.coef_)
print(mean_squared_error(y_test,prediction))
```

### 229

Consider variables x and y created from a pandas dataframe “car.csv”.
Create new column named “Age_car” (Age_car=2023-year)

For multiple linear regression problem, x contains the independent variables (Age_car, Driven_kms, Fuel_Type, Selling_type, Transmission) and y contains the dependent (Selling_Price) variable which is to be predicted.

Write a Python program to split x and y into training and testing datasets with a 20% split. Then create a multiple linear regression model using the training data and print its coefficients, intercept and mean squared error.

---

```python
# Import libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
# Import Dataset
df = pd.read_csv("car data.csv")
df["Age_car"] = 2023 - df["Year"]
df.head(2)

# x contains the independent variables (Age_car, Driven_kms, Fuel_Type, Selling_type, Transmission) 
# and y contains the dependent (Selling_Price)
x = df.drop(columns=["Car_Name","Year","Selling_Price","Present_Price","Owner"])
y = df["Selling_Price"]
print(x.shape)
print(y.shape)
x = pd.get_dummies(x)

x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,random_state=42
)

print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)

model = LinearRegression()
model.fit(x_train,y_train)
predictions = model.predict(x_test)
print(predictions)
print(model.coef_)
print(model.intercept_)
print(mean_squared_error(y_test,predictions))
```

### 230

Write a python program for following tasks -

Make sure all the required libraries are imported at the beginning of the Jupyter Notebook.

a) Load the windpower dataset into the notebook.
b) Check records for any missing values and remove them permanently using appropriate method.
c) Treat windspeed as the independent variable and power as the dependent variable. Split the dataset into training (75%) and testing (25%) sets using random state of 42.
d) Use scikit-learn to train polynomial regression models of degrees 3, 4, 5, and 6.
e) Predict power values on the test set and print the Mean Squared Error (MSE) and R² score (coefficient of determination) for each polynomial degree.
f) Plot actual vs. predicted power values for all polynomial degrees on the same graph for comparison and label the plot accordingly.
g) Create a plot showing MSE vs. polynomial degree. Which degree has the lowest MSE? Write in the comment.
h) Print the coefficients and intercept for each trained polynomial regression model.
i) Use each model to predict power output for wind speeds of 5, 10, 15, and 20 m/s, and print the predictions.

---

```python
# a) Load the windpower dataset into the notebook
import pandas as pd
df = pd.read_csv("windpower.csv")
df.head(2)
```

```python
# b) Check records for any missing values and remove them permanently using appropriate method.
df.isna().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
#  Treat windspeed as the independent variable and power as the dependent variable. Split the dataset into 
# training (75%) and testing (25%) sets using random state of 42.
x = df[["Wind speed (m/s)"]]
y = df["Power (kW)"]
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.25,random_state=42
)
```

```python
# Polynomial regression for degrees 3,4,5,6
degrees = [3,4,5,6]
msevalues = []
r2scores = []
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error,r2_score
plt.style.use('ggplot')
plt.figure(figsize=(12,8))
plt.title("Polynomial regression of windpower")
plt.scatter(x,y,color='k',label='truevalues')
## PREDICTIONS
windspeeds = np.array([[5],[10],[15],[20]])
windpredictions = {}
for degree in degrees:
    poly = PolynomialFeatures(degree = degree)
    x_train_poly = poly.fit_transform(x_train)
    x_test_poly = poly.fit_transform(x_test)
    model = LinearRegression()
    model.fit(x_train_poly,y_train)
    predictions = model.predict(x_test_poly)
    r2score = r2_score(y_test,predictions)
    r2scores.append(r2score)
    mse = mean_squared_error(y_test,predictions)
    msevalues.append(mse)
    print(f"Result for {degree} polynomial regression:")
    print("Coefficients = ",model.coef_)
    print("Intercept = ",model.intercept_)
    print(f"MSE = {mse:.2f}")
    print(f"R2score = {r2score:.2f}")

    plt.plot(x_test,predictions,'x',
             label=f"Predicted values for {degree} degree")
    windpoly = poly.fit_transform(windspeeds)
    powerprediction = model.predict(windpoly)
    windpredictions[degree] = powerprediction
plt.legend()
plt.xlabel("Windspeed")
plt.ylabel("WindPower")
plt.show()
    
    
    
    
    
```

```python
## LINE PLOT OF MSE VS DEGREE
## 6th Degree has the lowest MSE
plt.plot(
    degrees,msevalues,marker='o',linestyle='-',color='b'
)
plt.title("MSE vs Degree")
plt.xlabel("Degree")
plt.ylabel("MSE")
plt.grid(True)
plt.show()
```

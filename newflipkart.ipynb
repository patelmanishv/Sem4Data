You are provided with an HTML file named `newflipkart.html` that contains listings of mobile phones from an e-commerce website. Each listing contains relevant product details such as:

* Name and specifications
* Star rating
* Number of reviews
* Original price
* Discounted price
* Discount percentage
* Exchange offer (if available)

Write a Python program using **BeautifulSoup** to extract this data for each phone and display the dataframe output as given below - 


Your program must be able to:

* Handle entries that are incomplete or missing certain fields.
* Convert price-related text data into appropriate numeric format.
* Strictly follow the format given in the output.
